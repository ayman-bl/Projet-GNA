# Expérience 1 : Récupération de la graine LCG

```python
import sys
import os
sys.path.append(os.path.abspath('..'))
from generators.lcg import LCG

a, c, m = 1664525, 1013904223, 2**32
graine_secrete = 7777 # La graine que l'on va essayer de retrouver

# la victime génère un nombre
victime_lcg = LCG(graine_secrete, a, c, m)
sortie_interceptee = victime_lcg.next()
print(f"Étape 1 : L'attaquant intercepte la valeur -> {sortie_interceptee}")

#on inverse la fonction
a_inv = pow(a, -1, m) 
graine_calculee = ((sortie_interceptee - c) * a_inv) % m

print(f"Étape 2 : Graine calculée par notre script -> {graine_calculee}")
print(f"Conclusion : L'attaque a-t-elle réussi ? -> {graine_secrete == graine_calculee}")
```
