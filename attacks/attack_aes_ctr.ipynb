# Expérience 2 : Vulnérabilité liée à la réutilisation d'un nonce en AES-CTR

```python
import os
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.backends import default_backend

cle = os.urandom(32)
nonce_vulnerable = os.urandom(16) 

msg1 = b"message secret "
msg2 = b"texte tres prive"

# chiffrement des deux messages par la victime
cipher = Cipher(algorithms.AES(cle), modes.CTR(nonce_vulnerable), backend=default_backend())
c1 = cipher.encryptor().update(msg1) + cipher.encryptor().finalize()
c2 = cipher.encryptor().update(msg2) + cipher.encryptor().finalize()

# XOR des chiffrés
c1_xor_c2 = bytes(a ^ b for a, b in zip(c1, c2))
m1_xor_m2 = bytes(a ^ b for a, b in zip(msg1, msg2))

# résultat
print(f"L'attaque a-t-elle réussi (C1^C2 == M1^M2) ? : {c1_xor_c2 == m1_xor_m2}")
```
